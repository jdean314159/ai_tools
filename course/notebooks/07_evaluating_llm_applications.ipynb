# 07 — Evaluating LLM Applications

This notebook is the canonical teaching walkthrough for comparing **baseline** and **augmented** behavior.

Use it after the memory and RAG stages to answer a concrete question: did the added system component actually improve the result?


## Learning goals

By the end of this notebook you should be able to:
- separate baseline scoring from augmented scoring
- choose between substring, similarity, rubric, and judge-based evaluation
- record evidence and rationale instead of only pass/fail
- explain why an apparent improvement may still be misleading


```python
from llm_harness_core import EvaluatorRequest, SimilarityEvaluator, SubstringMatchEvaluator

baseline_answer = 'The project uses SQLite for analytics.'
grounded_answer = 'The project uses DuckDB for analytics, and the evidence comes from the updated analytics decision note.'

answer_eval = SubstringMatchEvaluator().evaluate(
    EvaluatorRequest(
        candidate=grounded_answer,
        expected_texts=('duckdb',),
        forbidden_texts=('sqlite for analytics',),
        min_expected_hits=1,
        max_forbidden_hits=0,
    )
).value

similarity_eval = SimilarityEvaluator(threshold=0.6).evaluate(
    EvaluatorRequest(
        candidate=grounded_answer,
        reference_answer='DuckDB is the current analytics database choice.',
    )
).value

answer_eval, similarity_eval
```
